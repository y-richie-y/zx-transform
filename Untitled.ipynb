```python
# QFT
n = 4
c = Circuit(n)

for i in range(n):
    for j in range(n - i):
        if j == 0:
            c.add_gate(H(i))
        else:
            c.add_gate(CRz(pi / 2 ** j, i + j, i))

c._to_svg()
```
