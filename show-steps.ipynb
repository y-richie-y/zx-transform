```python
from pyzx import draw
from pyzx import full_reduce
from pyzx.simplify import full_reduce_iter
from pyzx.generate import cliffords

from copy import deepcopy

from fractions import Fraction

def check(g1, g2):
    p1, p2 = g1.phases(), g2.phases()
    t1, t2 = g1.types(), g2.types()
    e1, e2 = sorted(g1.edges()), sorted(g2.edges())
    return p1 == p2 and t1 == t2 and e1 == e2

def filter_steps(steps):
    old_g = None
    for g, desc in steps:
        if old_g is None or not check(old_g, g):
            yield g, desc
        old_g = deepcopy(g)

def show_steps(gg):
    gs = []
    for g, desc in filter_steps(full_reduce_iter(gg)):
        print('desc', desc)
        draw(g)
        gs.append(deepcopy(g))
    return gs
```

```python
c = Circuit(1)
c.add_gate("ZPhase", 0, 0.3)
c.add_gate("NOT", 0)
c.add_gate("ZPhase", 0, 0.3)
c = c.to_graph()

draw(c)
steps = show_steps(c)
```
